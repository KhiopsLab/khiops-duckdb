# Train from Parquet

Create a small Parquet file, then train with the high-level API.

```python
from pathlib import Path

import duckdb

workdir = Path("_build/demo")
workdir.mkdir(parents=True, exist_ok=True)
parquet_path = workdir / "observations.parquet"
with duckdb.connect() as con:
    con.read_csv(
        "datasets/observations.csv",
        dtype={"id": "VARCHAR", "target": "VARCHAR"},
    ).write_parquet(str(parquet_path))
```

```python
from khiops_duckdb import train_predictor

report, model = train_predictor(
    parquet_path,
    target="target",
    output_dir=workdir / "results",
    max_trees=0,
    max_cores=1,
)
assert Path(report).is_file()
assert model is not None and Path(model).is_file()
{"report": report, "model": model}
```
